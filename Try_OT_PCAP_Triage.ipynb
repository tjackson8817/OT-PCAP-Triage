No install required. Run each cell below in order (click the little play button on the left of each cell, or press Shift+Enter).

**Step 1:** Run the setup cell below (installs the tool -- takes about 20 seconds).

**Step 2:** Run the upload cell and choose a `.pcap` or `.pcapng` file from your computer.

**Step 3:** Run the report cell to see your results and get a download link.

```
# Step 1: Setup (run this first)
!pip install -q "ot-pcap-triage @ git+https://github.com/tjackson8817/OT-PCAP-Triage.git"
print("Setup complete.")
```

```
# Step 2: Upload your capture file
from google.colab import files
uploaded = files.upload()
pcap_filename = list(uploaded.keys())[0]
print(f"Uploaded: {pcap_filename}")
```

```
# Step 3: Run the triage and view the report
from ot_pcap_triage import triage, render_markdown, build_summary_rows
from IPython.display import Markdown, display

findings = triage(pcap_filename)
report_md, summary_rows = render_markdown(findings, pcap_filename)

out_path = pcap_filename + ".report.md"
with open(out_path, "w") as f:
    f.write(report_md)

display(Markdown(report_md))

print("\n\nDownloading your report file now...")
files.download(out_path)
```
